```python
import os
import pandas as pd
import pyodbc
import warnings
warnings.filterwarnings("ignore", category=UserWarning)

DB_PATH = r"D:\Portfolio\nysed-analysis\data\raw\SRC2025_Group4.accdb"
```

```python
conn_str = (
    r"DRIVER={Microsoft Access Driver (*.mdb, *.accdb)};"
    rf"DBQ={DB_PATH};"
)

conn = pyodbc.connect(conn_str)
cursor = conn.cursor()
```

```python
# Get row and column counts for each table
tables = [row.table_name for row in cursor.tables(tableType="TABLE")]

for table in tables:
    cursor.execute(f"SELECT * FROM [{table}]")
    col_count = len(cursor.description)
    row_count = len(cursor.fetchall())
    print(f"{table}: {row_count} rows, {col_count} columns")
```

```python
# Folder to save raw table exports
PROCESSED_DIR = "../data/processed"
os.makedirs(PROCESSED_DIR, exist_ok=True)

for table in tables:
    df = pd.read_sql(f"SELECT * FROM [{table}]", conn)
    filename = table.replace(" ", "_").replace("/", "-") + ".csv"
    df.to_csv(f"{PROCESSED_DIR}/{filename}", index=False)
    print(f"Saved {filename}: {df.shape[0]} rows, {df.shape[1]} cols")
```
